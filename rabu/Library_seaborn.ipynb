```python
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('tips')
```

```python
sns.histplot() 

```

```python
sns.histplot(
    data=df,
    x='total_bill',  # Kolom data yang ingin dibuat histogramnya
    kde=True,  # Menambahkan garis kurva estimasi densitas (KDE)
    bins=20,  # Jumlah kotak/bin pada histogram
    color='skyblue',  # Warna batang histogram
)
```

```python
sns.boxplot(
    data=df,
    x='day',  # Variabel kategori pada sumbu X (contoh: hari)
    y='total_bill',  # Variabel numerik pada sumbu Y (contoh: total tagihan)
    hue='sex',  # Membagi data lebih lanjut berdasarkan kategori (opsional)
    palette='Set2',  # Memilih tema warna
)
```

```python
sns.scatterplot(
    data=df,
    x='total_bill',  # Variabel pada sumbu X (contoh: total tagihan)
    y='tip',         # Variabel pada sumbu Y (contoh: besar tip)
    hue='sex',       # Membedakan warna titik berdasarkan kategori (jenis kelamin)
    size='size',     # Mengubah ukuran titik berdasarkan jumlah orang di meja
    sizes=(20, 200), # Rentang ukuran minimum dan maksimum titik
    palette='deep',  # Tema warna
    alpha=0.7        # Tingkat transparansi titik (0 = transparan, 1 = solid)
)
```

```python
corr_matrix = df.select_dtypes(include=['float64', 'int64']).corr()
```

```python
plt.figure(figsize=(8, 6))

# 4. Membuat heatmap menggunakan sns.heatmap()
sns.heatmap(
    data=corr_matrix,
    annot=True,       # Menampilkan angka/nilai korelasi di dalam setiap kotak
    cmap='coolwarm',  # Tema warna (pilihan lain: 'viridis', 'Blues', 'YlGnBu')
    fmt='.2f',        # Format angka desimal (2 angka di belakang koma)
    linewidths=0.5,   # Ketebalan garis pemisah antar kotak
    cbar=True,        # Menampilkan bar skala warna di sebelah kanan
    vmin=-1, vmax=1   # Batas nilai minimum dan maksimum skala warna korelasi
)
```

```python
sns.pairplot(
    data=df,
    hue='sex',        # Membedakan warna titik/kurva berdasarkan kategori tertentu
    palette='husl',   # Tema warna untuk kategori
    diag_kind='kde',  # Jenis grafik pada diagonal utama ('kde' untuk kurva densitas, 'hist' untuk histogram)
    corner=True       # Menampilkan hanya bagian bawah matriks (menghindari duplikat grafik yang sama)
)
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Membuat data contoh (terdapat nilai 100 yang merupakan outlier/pencilan)
data = {'nilai': [10, 12, 14, 15, 13, 11, 100, 12, 14, 13]}
df = pd.DataFrame(data)

# 2. Menghitung z-score menggunakan stats.zscore() pada kolom tertentu
df['z_score'] = stats.zscore(df['nilai'])

# 3. Menampilkan DataFrame lengkap dengan nilai Z-score-nya
print("DataFrame dengan Kolom Z-Score:")
print(df)

# 4. Contoh Aplikasi: Menyaring Outlier
# Biasanya data dianggap sebagai outlier jika nilai absolut Z-score > 3 (atau > 2.5 untuk data kecil)
threshold = 2.5
outliers = df[abs(df['z_score']) > threshold]

print("\nData yang terdeteksi sebagai outlier (Pencilan):")
print(outliers)
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Membuat data contoh (kita buat data yang condong ke kanan/positif dengan adanya nilai besar)
data = [10, 12, 14, 15, 13, 11, 12, 14, 13, 50, 75]
df = pd.DataFrame({'nilai': data})

# 2. Menghitung nilai skewness menggunakan stats.skew()
# Parameter 'bias=False' (default) menggunakan koreksi sampel (bias-corrected), 
# mirip dengan perhitungan pada Pandas/Excel.
nilai_skew = stats.skew(df['nilai'])

print(f"Nilai Skewness: {nilai_skew:.4f}")

# 3. Interpretasi Sederhana Nilai Skewness:
if nilai_skew > 0.5:
    print("Interpretasi: Distribusi data condong ke kanan (Right-skewed / Positive skew).")
elif nilai_skew < -0.5:
    print("Interpretasi: Distribusi data condong ke kiri (Left-skewed / Negative skew).")
else:
    print("Interpretasi: Distribusi data relatif simetris (Normal).")
```

```python
import numpy as np

# 1. Membuat data contoh yang mengandung nilai NaN (Not a Number) dan Infinite (Inf)
data = np.array([10, 20, np.nan, 40, 50, np.inf])
print("Data Asli:", data)

# ==========================================
# 2. np.nan_to_num()
# ==========================================
# Berfungsi untuk membersihkan/mengubah nilai NaN menjadi 0 
# dan nilai Infinite (Inf) menjadi angka berhingga tertentu.
cleaned_data = np.nan_to_num(data, nan=0.0, posinf=100.0)
print("\n[np.nan_to_num] Data Setelah Dibersihkan:")
print(cleaned_data)

# ==========================================
# 3. np.mean() & np.std()
# ==========================================
# np.mean() untuk mencari rata-rata
# np.std() untuk mencari standar deviasi
mean_val = np.mean(cleaned_data)
std_val = np.std(cleaned_data)

print(f"\n[np.mean] Rata-rata: {mean_val:.2f}")
print(f"[np.std] Standar Deviasi: {std_val:.2f}")

# (Catatan: Jika Anda tidak ingin membersihkan data dulu, 
# Anda bisa menggunakan np.nanmean() dan np.nanstd() untuk mengabaikan NaN secara otomatis)

# ==========================================
# 4. np.linspace()
# ==========================================
# Berfungsi untuk menghasilkan sekumpulan angka yang berjarak sama persis 
# dalam rentang tertentu. Sangat berguna untuk membuat sumbu X pada grafik fungsi.
# Sintaks: np.linspace(start, stop, jumlah_titik)

x_vals = np.linspace(0, 10, 5)  # Membuat 5 titik dari 0 sampai 10
print("\n[np.linspace] Hasil:")
print(x_vals)  # Output: [ 0.   2.5  5.   7.5 10. ]
```
