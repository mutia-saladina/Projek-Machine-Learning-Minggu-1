```python
df['Description'] = df['Description'].fillna('Unknown')
```

```python
median_price = df['UnitPrice'].median()
df['UnitPrice'] = df['UnitPrice'].fillna(median_price)
```

```python
plt.figure(figsize=(10, 5))
df['Quantity'].hist(bins=50, color='steelblue', edgecolor='black')
plt.title('Distribusi Quantity Pembelian')
plt.xlabel('Quantity')
plt.ylabel('Frekuensi')
plt.show()
```

```python
df_clean = df.dropna(subset=['CustomerID'])
df_clean.shape
```

```python
df = df.drop(columns=['CustomerID'])
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df_clean['UnitPrice'])
plt.title('Boxplot Harga Satuan')
plt.show()
```

```python
Q1 = df_clean['UnitPrice'].quantile(0.25)
Q3 = df_clean['UnitPrice'].quantile(0.75)
IQR = Q3 - Q1
low = Q1 - 1.5 * IQR
high = Q3 + 1.5 * IQR

df_filtered = df_clean[(df_clean['UnitPrice'] >= low) & (df_clean['UnitPrice'] <= high)]
print("Shape sebelum filtering: ", df_clean.shape)
print("Shape sesudah filtering: ", df_filtered.shape)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(df_clean['Quantity'], df_clean['UnitPrice'], alpha=0.3)
plt.xlabel('Quantity')
plt.ylabel('Unit Price')
plt.title('Quantity vs UnitPrice')
plt.show()
```
